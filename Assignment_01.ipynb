We can see by pulling the data type of all the values from the DMERC that the type values range from object, int, and floats. What we will be looking at specifically is the object type values and our goal is to convert those values into float type variables.

To do this, we will utilize a panda function that

```python
objec_colm = df.select_dtypes(include='object').columns
print(objec_colm)
int_colm = df.select_dtypes(include='int64').columns
print(int_colm)
```

```python
for col in objec_colm:
 df[col] = df[col].astype(str).str.replace(',', '').str.strip()
 df[col] = pd.to_numeric(df[col], errors='coerce')

for col in int_colm:
 df[col] = df[col].astype(str)
```
